```scala
%spark

import spark.implicits._
import org.apache.spark.sql.{DataFrame, Dataset, Row}
import org.apache.spark.sql.functions._
import org.apache.spark.rdd.RDD
import org.apache.spark.storage.StorageLevel
```

```scala
%spark

val data = Seq(
  ("A", "B"),
  ("B", "C"),
  ("B", "D"),
  ("D", "E"),
  ("F", "G"),
  ("G", "H")
)

val dfEdges = data.toDF("src", "dst")
dfEdges.show(false)
```

```scala
%spark

def makeUndirected(dfEdges: DataFrame): DataFrame = {
  val cleaned = dfEdges
    .filter(col("src").isNotNull && col("dst").isNotNull)
    .filter(col("src") =!= col("dst"))
    .distinct()

  val reversed = cleaned.select(
    col("dst").as("src"),
    col("src").as("dst")
  )

  cleaned
    .union(reversed)
    .distinct()
}

val dfUndirected = makeUndirected(dfEdges).cache()
dfUndirected.show(false)
```

```scala
%spark

val nodesDF = dfUndirected
  .select(col("src").as("node"))
  .union(dfUndirected.select(col("dst").as("node")))
  .distinct()
  .cache()

nodesDF.show(false)
```

```scala
%spark

val undirectedRDD: RDD[(String, String)] = dfUndirected.rdd
  .map(r => (r.getString(0), r.getString(1)))
  .cache()

undirectedRDD.collect().foreach(println)
```

```scala
%spark

def runCCFRDD(dfUndirected: DataFrame): (RDD[(String, String)], Int) = {
  val undirectedRDD = dfUndirected.rdd
    .map(r => (r.getString(0), r.getString(1)))
    .cache()

  // key by destination so we can propagate neighbor labels back to src
  val graphByDstRDD = undirectedRDD
    .map { case (src, dst) => (dst, src) }
    .cache()

  var labelsRDD = undirectedRDD
    .flatMap { case (a, b) => Seq(a, b) }
    .distinct()
    .map(node => (node, node))
    .cache()

  var changed = true
  var iteration = 0

  while (changed && iteration < 100) {
    iteration += 1

    val propagated = graphByDstRDD
      .join(labelsRDD) // (dst, (src, label_of_dst))
      .map { case (_, (src, labelOfNeighbor)) => (src, labelOfNeighbor) }

    val minNeighborLabels = propagated
      .reduceByKey((a, b) => if (a < b) a else b)

    val nextLabelsRDD = labelsRDD
      .leftOuterJoin(minNeighborLabels)
      .mapValues {
        case (currentLabel, Some(neighborMin)) =>
          if (neighborMin < currentLabel) neighborMin else currentLabel
        case (currentLabel, None) =>
          currentLabel
      }
      .cache()

    val numChanges = labelsRDD
      .join(nextLabelsRDD)
      .filter { case (_, (oldLabel, newLabel)) => oldLabel != newLabel }
      .count()

    changed = numChanges > 0

    labelsRDD.unpersist()
    labelsRDD = nextLabelsRDD
  }

  (labelsRDD, iteration)
}
```

```scala
%spark

val (finalLabelsRDD, rddIterations) = runCCFRDD(dfUndirected)

println(s"RDD iterations = $rddIterations")

finalLabelsRDD
  .collect()
  .sortBy(_._1)
  .foreach(println)
```

```scala
%spark

spark.sparkContext.setCheckpointDir("/tmp/spark-checkpoints")
```

```scala
%spark

def runCCFDataFrame(dfUndirected: DataFrame): (DataFrame, Int) = {
  val nodesDF = dfUndirected
    .select(col("src").as("node"))
    .union(dfUndirected.select(col("dst").as("node")))
    .distinct()

  var labelsDF = nodesDF
    .withColumn("label", col("node"))
    .persist(StorageLevel.MEMORY_AND_DISK)

  labelsDF.count()

  var changed = true
  var iteration = 0

  while (changed && iteration < 100) {
    iteration += 1

    val propagated = dfUndirected
      .join(labelsDF, dfUndirected("dst") === labelsDF("node"))
      .select(
        dfUndirected("src").as("node"),
        labelsDF("label").as("neighbor_label")
      )

    val minNeighborLabels = propagated
      .groupBy("node")
      .agg(min("neighbor_label").as("min_neighbor_label"))

    val nextLabelsRaw = labelsDF
      .join(minNeighborLabels, Seq("node"), "left")
      .select(
        col("node"),
        least(col("label"), coalesce(col("min_neighbor_label"), col("label"))).as("label")
      )

    val nextLabelsDF = nextLabelsRaw.localCheckpoint(eager = true)

    val oldDf = labelsDF.alias("old")
    val newDf = nextLabelsDF.alias("new")

    val numChanges = oldDf
      .join(newDf, Seq("node"))
      .filter(col("old.label") =!= col("new.label"))
      .count()

    changed = numChanges > 0

    labelsDF.unpersist()
    labelsDF = nextLabelsDF.persist(StorageLevel.MEMORY_AND_DISK)
    labelsDF.count()
  }

  (labelsDF, iteration)
}
```

```scala
%spark

val (finalLabelsDF, dfIterations) = runCCFDataFrame(dfUndirected)

println(s"DataFrame iterations = $dfIterations")

finalLabelsDF
  .select(col("node"), col("label").as("component"))
  .orderBy("node")
  .show(false)
```

```scala
%spark

def generateMultiComponentGraph(numComponents: Int, componentSize: Int): DataFrame = {
  val edges = (0 until numComponents).flatMap { c =>
    val start = c * componentSize + 1
    val end = start + componentSize - 1
    (start until end).map(i => (i.toString, (i + 1).toString))
  }
  edges.toDF("src", "dst")
}
```

```scala
%spark

def benchmarkRDD(dfUndirected: DataFrame): (Long, Int) = {
  val start = System.nanoTime()
  val (_, iterations) = runCCFRDD(dfUndirected)
  val end = System.nanoTime()
  val durationMs = (end - start) / 1000000
  (durationMs, iterations)
}
```

```scala
%spark

def benchmarkDataFrame(dfUndirected: DataFrame): (Long, Int) = {
  val start = System.nanoTime()
  val (_, iterations) = runCCFDataFrame(dfUndirected)
  val end = System.nanoTime()
  val durationMs = (end - start) / 1000000
  (durationMs, iterations)
}
```

```scala
%spark

val testConfigs = Seq(
  (5, 10),     // 50 nodes
  (5, 20),   // 100 nodes
  (10, 20),   // 200 nodes
  (10, 40)    // 400 nodes
)
```

```scala
%spark

val benchmarkResults = testConfigs.map { case (numComponents, componentSize) =>
  val dfGraph = generateMultiComponentGraph(numComponents, componentSize)
  val dfUndirectedBench = makeUndirected(dfGraph).cache()
  dfUndirectedBench.count()

  val numNodes = numComponents * componentSize
  val numEdges = dfGraph.count()

  val (rddTimeMs, rddIterations) = benchmarkRDD(dfUndirectedBench)
  val (dfTimeMs, dfIterations) = benchmarkDataFrame(dfUndirectedBench)

  dfUndirectedBench.unpersist()

  (
    numNodes,
    numEdges,
    rddTimeMs,
    rddIterations,
    dfTimeMs,
    dfIterations
  )
}
```

```scala
%spark

val resultsDF = benchmarkResults.toDF(
  "num_nodes",
  "num_edges",
  "rdd_time_ms",
  "rdd_iterations",
  "df_time_ms",
  "df_iterations"
)

resultsDF.show(false)
```
